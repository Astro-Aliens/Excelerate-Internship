```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("RIT+Opportunity+Wise+Data+-+Sheet1.csv")

print("Original Shape:", df.shape)

# --------------------------------------------------
# 1. Remove Exact Duplicate Rows
# --------------------------------------------------
df = df.drop_duplicates()

# --------------------------------------------------
# 2. Remove Leading/Trailing Spaces
# --------------------------------------------------
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype(str).str.strip()

# --------------------------------------------------
# 3. Replace Blank Strings with NaN
# --------------------------------------------------
df.replace(r'^\s*$', np.nan, regex=True, inplace=True)

# --------------------------------------------------
# 4. Standardize Gender Values
# --------------------------------------------------
if 'Gender' in df.columns:
    df['Gender'] = (
        df['Gender']
        .str.title()
        .replace({
            'Male ': 'Male',
            'Female ': 'Female'
        })
    )

# --------------------------------------------------
# 5. Standardize Country Names
# --------------------------------------------------
if 'Country' in df.columns:
    df['Country'] = df['Country'].str.title()

# --------------------------------------------------
# 6. Convert Date Columns
# --------------------------------------------------
date_columns = [
    'Learner SignUp DateTime',
    'Opportunity End Date',
    'Date of Birth',
    'Entry created at',
    'Apply Date',
    'Opportunity Start Date'
]

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')

# --------------------------------------------------
# 7. Fill Missing Text Fields
# --------------------------------------------------
text_columns = [
    'Institution Name',
    'Current/Intended Major'
]

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].fillna('Unknown')

# --------------------------------------------------
# 8. Remove Fully Empty Rows
# --------------------------------------------------
df = df.dropna(how='all')

# --------------------------------------------------
# 9. Remove Duplicate Opportunity Records
# (Optional Safety Check)
# --------------------------------------------------
df = df.drop_duplicates()

# --------------------------------------------------
# 10. Check Remaining Missing Values
# --------------------------------------------------
print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nCleaned Shape:")
print(df.shape)

# --------------------------------------------------
# 11. Save Cleaned Dataset
# --------------------------------------------------
df.to_csv("RIT_Cleaned_Dataset_Final.csv", index=False)

print("\nCleaned dataset saved successfully!")
print("File Name: RIT_Cleaned_Dataset_Final.csv")
```
