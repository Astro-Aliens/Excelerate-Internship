```python
# WEEK 1 DELIVERABLE - DATA UNDERSTANDING & PREPARATION
# AI Powered Data Analysis Internship
# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Display all columns
pd.set_option('display.max_columns', None)

# 2. Load Dataset

df = pd.read_csv("RIT+Opportunity+Wise+Data+-+Sheet1.csv")

print("="*60)
print("DATASET LOADED SUCCESSFULLY")
print("="*60)
```

```python
# 3. Dataset Understanding

print("\nDATASET DIMENSIONS")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nCOLUMN NAMES")
for col in df.columns:
    print(col)

print("\nDATA TYPES")
print(df.dtypes)

print("\nDATASET INFORMATION")
print(df.info())

# First 5 rows
print("\nFIRST 5 RECORDS")
display(df.head())
```

```python
# 4. Data Dictionary

print("\nDATA DICTIONARY")
data_dict = pd.DataFrame({
    'Column Name': df.columns,
    'Data Type': df.dtypes.values
})

display(data_dict)
```

```python
# ----------------------------
# 5. Missing Values Analysis
# ----------------------------
missing_df = pd.DataFrame({
    'Missing Values': df.isnull().sum()
})

missing_df['Percentage'] = (
    missing_df['Missing Values'] / len(df) * 100
).round(2)

display(missing_df)
```

```python
# 6. Duplicate Records Check

duplicates = df.duplicated().sum()

print("\nDUPLICATE RECORDS")
print("Total Duplicate Records:", duplicates)
```

```python
# 7. Data Quality Validation

print("\nUNIQUE OPPORTUNITY CATEGORIES")
print(df['Opportunity Category'].value_counts())

print("\nUNIQUE STATUS DESCRIPTIONS")
print(df['Status Description'].value_counts())

print("\nGENDER DISTRIBUTION")
print(df['Gender'].value_counts())
```

```python
# 8. Summary Statistics

print("\nSUMMARY STATISTICS")
display(df.describe(include='all'))

# 9. Top Opportunity Categories

print("\nTOP OPPORTUNITY CATEGORIES")
category_counts = df['Opportunity Category'].value_counts()
display(category_counts)
```

```python
#10. Gender Participation

print("\nGENDER PARTICIPATION")
gender_counts = df['Gender'].value_counts()
display(gender_counts)

#11. Country Participation

print("\nTOP 10 COUNTRIES")
country_counts = df['Country'].value_counts().head(10)
display(country_counts)
```

```python
#12. Institution Participation

print("\nTOP 10 INSTITUTIONS")
institution_counts = df['Institution Name'].value_counts().head(10)
display(institution_counts)

#13. Visualization 1
# Opportunity Categories

plt.figure(figsize=(8,5))
df['Opportunity Category'].value_counts().plot(kind='bar')
plt.title('Opportunity Category Distribution')
plt.xlabel('Opportunity Category')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
#14. Visualization 2
# Gender Distribution

plt.figure(figsize=(7,7))
df['Gender'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title('Gender Distribution')
plt.ylabel('')
plt.show()

#15. Visualization 3
# Top Countries

plt.figure(figsize=(10,5))
df['Country'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Countries by Participation')
plt.xlabel('Country')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
#16. Visualization 4
# Top Opportunity Names

plt.figure(figsize=(14,12))
df['Opportunity Name'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Opportunities')
plt.xlabel('Opportunity Name')
plt.ylabel('Participants')
plt.tight_layout()
plt.show()

#17. Final Observations

print("\nFINAL OBSERVATIONS")
print("- Dataset successfully loaded and analyzed")
print("- Missing values identified")
print("- Duplicate records checked")
print("- Opportunity categories analyzed")
print("- Participation trends analyzed")
print("- Visualizations generated successfully")
```
